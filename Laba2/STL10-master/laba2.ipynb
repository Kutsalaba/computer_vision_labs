```python
from nltk.corpus import stopwords

# remove english stopwords function
def remove_stopwords(text, language):
    stop_words = set(stopwords.words(language))
    word_tokens = text.split()
    filtered_text = [word for word in word_tokens if word not in stop_words]
    print(language)
    print(filtered_text)
 
en_text = "This is a sample sentence and we are going to remove the stopwords from this"
remove_stopwords(en_text, "english")

tr_text = "bu cümledeki engellenen kelimeleri kaldıracağız"
remove_stopwords(tr_text, "turkish")
```

```python


# Define Turing Machine state transitions
def turing_machine(state, tape, head_position):
    if state == 'q1':
        current_symbol = tape[head_position]
        if current_symbol == '1':
            tape[head_position] = '0'
            return turing_machine('q2', tape, head_position + 1)
        else:
            return tape, head_position, 'q0'
    elif state == 'q2':
        current_symbol = tape[head_position]
        if current_symbol == '1':
            tape[head_position] = '0'
            return turing_machine('q0', tape, head_position + 1)
        else:
            return tape, head_position, 'q0'
    else:
        return tape, head_position, 'q0'

def process_tape(tape):
    # Initial state q1, starting head position at index 1
    initial_state = 'q1'
    head_position = 1  # Starting head position
    tape_list = list(tape)

    # Apply the Turing machine process
    final_tape, final_head_position, final_state = turing_machine(initial_state, tape_list, head_position)

    return ''.join(final_tape)

# Initial tape configuration (example input)
tape = '01101011110'

# Process the tape with the Turing machine
result = process_tape(tape)

print("Final tape configuration:", result)
```
